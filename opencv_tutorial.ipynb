##GrayScale Conversion

```python
img_grayscale=cv2.cvtColor(img_present,cv2.COLOR_RGB2GRAY)
display(img_grayscale)
```

```python
l=[i for i in dir(cv2) if i.startswith('COLORMAP_')]
flags=[getattr(cv2,i) for i in l]
d={}
for i in range(len(l)):
  d[l[i]]=flags[i]
for key, value in sorted(d.items(), key=lambda item: item[1]):
    print(f"{value:2d} : {key}")
```

```python
for i in l:
  flag=getattr(cv2,i)
  cmap_img=cv2.applyColorMap(img_grayscale,flag)
  print(f"Colormap: {i}")
  cmap_img=cv2.cvtColor(cmap_img,cv2.COLOR_BGR2RGB)
  display(cmap_img)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img_bgr = cv2.imread('/content/dog.jpg')
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

# 1. Convert to Grayscale & Blur
gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
blurred = cv2.GaussianBlur(gray, (7, 7), 0)

# 2. Use regular Otsu thresholding (without INV) to separate subject from light background
_, thresh = cv2.threshold(blurred, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

# 3. Find external contours
contours, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

output_img = img_rgb.copy()
img_h, img_w = gray.shape

for cnt in contours:
    area = cv2.contourArea(cnt)
    x, y, w, h = cv2.boundingRect(cnt)

    # Ignore contours that cover almost the entire canvas (image border artifact)
    if area > 15000 and (w < img_w * 0.95 or h < img_h * 0.95):
        cv2.rectangle(output_img, (x, y), (x + w, y + h), (0, 255, 0), 3)

# Display results
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1); plt.imshow(thresh, cmap='gray'); plt.title('Threshold Mask')
plt.subplot(1, 2, 2); plt.imshow(output_img); plt.title('Detected Bounding Box')
plt.axis('off')
plt.show()
```
